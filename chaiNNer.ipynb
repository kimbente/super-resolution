# Tensor to png

https://pytorch.org/vision/stable/generated/torchvision.utils.flow_to_image.html#torchvision.utils.flow_to_image

```python
wave_tensor = torch.load("/home/kim/super-resolution/torch_data/wavy_scene_60.pt")
```

```python
# Subset
img = wave_tensor[0, 0, :, :].unsqueeze(0) # torch.Size([1, 60, 60]

# Scale
img_min, img_max = img.min(), img.max()
new_min, new_max = 0, 255
img_p = (img - img_min)/(img_max - img_min) * (new_max - new_min) + new_min

# Copy
img_3c = torch.cat((img_p, img_p, img_p), dim = 0)

# Type
img_3c_byte = img_3c.type(torch.uint8)
img_3c_byte
```

```python
# Input needs to be torch.uint8 (Byte tensor)
torchvision.io.write_png(img_p.type(torch.uint8), 
                         filename = "/home/kim/super-resolution/torch_data/bed_wavy_image.png")

# Normalisation in function does not work perfectly
# torchvision.io.write_png(img.type(torch.uint8), 
#                         filename = "/home/kim/super-resolution/torch_data/bed_wavy_image.png")
```
