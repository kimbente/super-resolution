```python
import torch
from torch.utils.data import DataLoader

import numpy as np
import pandas as pd
import plotly.graph_objects as go
import plotly.express as px
from plotly.subplots import make_subplots

from torchmetrics.functional.image import image_gradients
```

# Exploring gradients 

```python
training_tensor = torch.load('./torch_data/TRANSANT_experiments/TRANSANT_training_tensor.pt')
```

```python
training_tensor[:, 1, :, :].unsqueeze(1).shape
```

```python
# Use image_gradients from torch metrics https://torchmetrics.readthedocs.io/en/stable/image/image_gradients.html
# Tuple of (dy, dx) with each gradient of shape [N, C, H, W]
dx, dy = image_gradients(training_tensor[:, 1, :, :].unsqueeze(1))
```

```python
fig = px.imshow(training_tensor[1, 0, :, :].squeeze(), 
                color_continuous_scale = 'RdBu_r',
                origin = "upper", 
                title = "Bed elevation")
fig.show()
```

```python
fig = px.imshow(training_tensor[1, 1, :, :].squeeze(), 
                color_continuous_scale = 'RdBu_r',
                origin = "upper", 
                title = "Surface elevation")
fig.show()
```

```python
fig = px.imshow(dx[1, :, :, :].squeeze(), 
                color_continuous_scale = 'RdBu_r',
                origin = "upper", 
                title = "gradients in dx direction")
fig.show()
```

```python
fig = px.imshow(dy[1, :, :, :].squeeze(), 
                color_continuous_scale = 'RdBu_r',
                origin = "upper", 
                title = "gradients in dx direction")
fig.show()
```
