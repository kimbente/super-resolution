## Import statements

```python
import matplotlib.pyplot as plt  # plotting library
# import numpy as np  # numerical library
# import xarray as xr  # netCDF library
# import cartopy  # Map projections libary
import cartopy.crs as ccrs  # Projections list
import cartopy.feature as cfeature # for coastlines
import matplotlib.patches as mpatches # draw domain boundry patches
```

### Defining span of domains

```python
# 22500m per scene

scene_size = 22500 # in meters
n_scenes = 30 # scenes per row/column: 900 scences per domain
span = scene_size * n_scenes
print(f"The span of each domain is {span} meters, consisting of {n_scenes} x {n_scenes} scences, witch each scene measuring 22.5 x 22.5 km.")
```

## Define two domains

When midpoints are aligned, the boundries of the datasets are still 25 meters apart.
BedMachine: 500m spacing so 250m to from midpoint to all four boundries. 
IceVelocity: 450 m spacing so 225m from midpoint to all four boundries.

Shift midpoint 4050 m to the right and 4075 m 

```python
# Transantarctic mountains: Nimrod, Byrd, Skelton glacier, Victoria land
# mountainous domain spanning grounded ice and floating ice.
# stick to order: y, x
transant_y_min = - 1232000
transant_x_min = 3500 # changed
transant_y_max = transant_y_min + span # -557000
transant_x_max = transant_x_min + span # 678500

# Dome C
# lake Vostok is still further north than this domain
domec_y_min = -1232000
domec_x_min = 899000
domec_y_max = domec_y_min + span # -557000
domec_x_max = domec_x_min + span # 1574000
```

## Visualise both domains

```python
# define font
hfont = {'fontname':'Helvetica'}
special_color = '#2D27EB'

# Initialise plot
fig = plt.figure(figsize = [10, 10])
ax = plt.axes(projection = ccrs.SouthPolarStereo())

# restrict to over 65 lat
ax.set_extent([-180, 180, -90, -65], ccrs.PlateCarree())

# hides boundry line
ax.axis('off')

# add grey land
ax.add_feature(cfeature.LAND, facecolor = ("#FAFAFA"), alpha = 1.0)

# thin coastline lines
ax.add_feature(cfeature.COASTLINE, edgecolor = special_color, linestyle = '-', linewidth = 0.4, alpha = 0.7)

# Dome C patch
ax.add_patch(mpatches.Rectangle(xy = [domec_x_min, domec_y_min], width = (domec_x_max - domec_x_min), height = (domec_y_max - domec_y_min),
                                facecolor = 'none', edgecolor = special_color, linewidth = 0.8,
                                transform = ccrs.SouthPolarStereo()))
# Transantarctic Mountains patch
ax.add_patch(mpatches.Rectangle(xy = [transant_x_min, transant_y_min], width = (transant_x_max - transant_x_min), height = (transant_y_max - transant_y_min),
                                facecolor = 'none', edgecolor = special_color, linewidth = 0.8,
                                transform = ccrs.SouthPolarStereo()))
# Add Dome C marker
ax.scatter(1359993, - 894443, 
         color = special_color, marker = "x",
         transform = ccrs.SouthPolarStereo())

# Add South Pole marker
ax.scatter(0, 0, 
         color = special_color, marker = "x",
         transform = ccrs.SouthPolarStereo())

# Add Dome C label
plt.annotate("Dome C", (1359993 - 300000, - 894443 + 100000), **hfont, color = special_color,
             transform = ccrs.SouthPolarStereo())

# Add South Pole label
plt.annotate("South Pole", (0 - 300000, 0 + 100000), **hfont, color = special_color,
             transform = ccrs.SouthPolarStereo())

plt.show()
```
