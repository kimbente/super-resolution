```python
import torch
import numpy as np
import pandas as pd
import plotly.graph_objects as go
import plotly.express as px
import xarray as xr
import matplotlib.pyplot as plt
```

# Load scene

Scene target tensor: [N, C, H, W]  

with channels: 
- `[:, 0, :, :]` bed
- `[:, 1, :, :]` surface
- `[:, 2, :, :]` thickness
- `[:, 3, :, :]` mask
- `[:, 4, :, :]` firn
- `[:, 5, :, :]` errorbed

and polar stereographic coordinates
- `[:, 6, :, :]` y
- `[:, 7, :, :]` x

```python
t = torch.load('./torch_data/TRANSANT_bed_scenes_60pixel.pt')
```

```python
torch.corrcoef(
    torch.cat((t[:, 0, :, :].reshape(1, -1),
               t[:, 1, :, :].reshape(1, -1)), dim = 0))
```

```python
values, counts = np.unique(t[:, 3, :, :].reshape(-1).numpy(), return_counts = True)
# ocean (0); ice-free land (1); grounded ice (2); floating ice (3);
np.round(counts/t[:, 3, :, :].reshape(-1).shape[0], decimals=3)
```

```python
fig = px.imshow(t[10, 2, :, :], 
                color_continuous_scale = 'RdBu_r',
                # DEFAULT for matrices: only tensor because of conversion
                origin = "upper", title = "Bed elevation [channel 0] of scene index 1")
fig.show()
```
