```python
import torch
import numpy as np
import pandas as pd
import plotly.graph_objects as go
import plotly.express as px
import gpytorch
import xarray as xr
```

# Artificially reduce resolution (magnify)

```python
# Tensor needs explicit dimension 1
d = torch.arange(1, 401).reshape(1, 20, -1)
print(d.shape)

# Magnify
magnification_factor = 2
magnify = torch.nn.AvgPool2d(kernel_size = magnification_factor)
# stride is my default the kernel size

d_mag = magnify(d)
print(d_mag.shape)

print(d[0, 0:2, 0:2])
print(d_mag[0, 0, 0])
```

```python
# permute to reshape it to (H, W, C)
img = d_mag.permute(1, 2, 0)
img = img.squeeze().detach().numpy()
fig = px.imshow(img, binary_string = True, 
origin = 'lower')
fig.update_xaxes(showticklabels = False)
fig.update_yaxes(showticklabels = False)
fig.show()
# https://stackoverflow.com/questions/38973868/adjusting-gridlines-and-ticks-in-matplotlib-imshow
```
